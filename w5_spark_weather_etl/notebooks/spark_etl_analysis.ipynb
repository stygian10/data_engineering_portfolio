```python
# ===============================================
# Week 5 Spark ETL Analysis – Combined Plots per City
# ===============================================

import matplotlib.pyplot as plt

cities = df_pd['city'].unique()
num_cities = len(cities)

fig, axes = plt.subplots(num_cities, 3, figsize=(18, 5*num_cities), sharex=True)

for i, city in enumerate(cities):
    city_data = df_pd[df_pd['city'] == city].sort_values('date')

    # Temperature & Rolling Avg
    axes[i,0].plot(city_data['date'], city_data['avg_temp'], label='Avg Temp', color='tab:blue')
    axes[i,0].plot(city_data['date'], city_data['rolling_avg_temp'], linestyle='--', label='Rolling Avg Temp', color='tab:orange')
    axes[i,0].set_title(f"{city} - Temperature")
    axes[i,0].set_ylabel("°C")
    axes[i,0].legend()
    axes[i,0].grid(True)

    # Humidity
    axes[i,1].plot(city_data['date'], city_data['avg_humidity'], color='tab:green')
    axes[i,1].set_title(f"{city} - Humidity")
    axes[i,1].set_ylabel("%")
    axes[i,1].grid(True)

    # Windspeed
    axes[i,2].plot(city_data['date'], city_data['avg_windspeed'], color='tab:red')
    axes[i,2].set_title(f"{city} - Wind Speed")
    axes[i,2].set_ylabel("m/s")
    axes[i,2].grid(True)

# Common X-axis label
for ax in axes[-1,:]:
    ax.set_xlabel("Date")
    for tick in ax.get_xticklabels():
        tick.set_rotation(45)

plt.tight_layout()
plt.suptitle("Weather Summary per City (Week 5 ETL)", fontsize=16, y=1.02)
plt.show()
```
